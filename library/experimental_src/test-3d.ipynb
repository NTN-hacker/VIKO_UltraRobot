```python
import os
import numpy as np
import plotly.graph_objects as go
import plotly.io as pio

# Directory paths
directory_path = '../laser/test_a_scan_laser/350'
output_directory = '../laser/view_a_scan/350'
os.makedirs(output_directory, exist_ok=True)

for filename in os.listdir(directory_path):
    if filename.endswith('.txt'):
        file_path = os.path.join(directory_path, filename)

        try:
            # Load data
            data = np.loadtxt(file_path, delimiter=' ', dtype=float)  # Use appropriate delimiter if needed
            if data.shape[1] != 2:
                raise ValueError("Data must contain exactly two columns for x and y coordinates.")
        except Exception as e:
            print(f"Error reading {filename}: {e}")
            continue

        # Split data into x and y
        x, y = data[:, 0], data[:, 1]

        # Create a scatter plot
        fig = go.Figure(data=go.Scatter(x=x, y=y, mode='lines'))

        fig.update_layout(
            title=f'Scatter Plot of Coordinates - {filename}',
            xaxis_title='X-axis',
            yaxis_title='Y-axis'
        )

        fig.show()

        # Save the figure as an HTML file
        html_path = os.path.join(output_directory, f"{os.path.splitext(filename)[0]}.html")
        pio.write_html(fig, file=html_path)

        print(f"Processed and saved: {filename}")
```

```python
import numpy as np
from scipy import stats

# Load the data from the file
data = np.loadtxt('E:\Quan\AutoRoboticInspection\VIKO_UltraRobot\laser\datascan-python_2024-07-18_12-35-41_2000l-5000e-3900i.txt')

# Calculate the mean and standard deviation of the data
mean = np.mean(data)
std_dev = np.std(data)

# Define the range based on the mean and standard deviation
lower_bound = mean - 2 * std_dev
upper_bound = mean + 2 * std_dev

# Filter out rows containing out-of-range values
data_cleaned = data[np.all((data >= lower_bound) & (data <= upper_bound), axis=1)]

# Find the most popular value in the cleaned data
if data_cleaned.size > 0:
    mode_result= stats.mode(data_cleaned, axis=None)
    mode_value = mode_result.mode[0] if mode_result.mode.size > 0 else mode_result.mode
    print(f"The most popular value is: {mode_value}")
else:
    print("No data left after cleaning out-of-range values.")
```
