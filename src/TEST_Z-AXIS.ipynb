```python
import plotly.express as px 
import cv2
import numpy as np
#CAMERA
from pypylon import pylon
def getImage(path: str)-> np.array:
    image = None
    camera = pylon.InstantCamera(pylon.TlFactory.GetInstance().CreateFirstDevice())

    camera.StartGrabbing(pylon.GrabStrategy_LatestImageOnly) 
    converter = pylon.ImageFormatConverter()

    converter.OutputPixelFormat = pylon.PixelType_BGR8packed
    converter.OutputBitAlignment = pylon.OutputBitAlignment_MsbAligned

    prev_frame_time = 0
    new_frame_time = 0

    index = 0
    while camera.IsGrabbing():
        grabResult = camera.RetrieveResult(5000, pylon.TimeoutHandling_ThrowException)

        if grabResult.GrabSucceeded():
            image1 = converter.Convert(grabResult)
            image = image1.GetArray()
            image = cv2.resize(image, (2448, 2048), fx=0.8366, fy=1, interpolation=cv2.INTER_LINEAR)
            cv2.imwrite(f'../data/Robotic/Distance_Z_Axis/test_{path}.png', image)
            break
        grabResult.Release()    
        
    camera.StopGrabbing()
    print('Closing Camera, Done')
    return image

#VISION
def detectPoint(img)-> list:
    x, y = None, None
    img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    res, thresh = cv2.threshold(img_gray, 90, 255, cv2.THRESH_BINARY)
    contours, hierarchy = cv2.findContours(image=thresh, mode=cv2.RETR_TREE, 
                                        method=cv2.CHAIN_APPROX_NONE)
    
    # draw contours on the original image
    image_copy = img.copy()
    for contour in contours:

        area = cv2.contourArea(contour)
        print(area)
        if 1500 < area < 3000:
            cv2.drawContours(image_copy, contours=contour, contourIdx=-1, 
                            color=(0, 255, 0), thickness=2, lineType=cv2.LINE_AA)
            M = cv2.moments(contour)
            if M['m00'] != 0:
                cx = float(M['m10'] / M['m00'])
                cy = float(M['m01'] / M['m00'])
                x, y, = cx, cy
                print(f"Coordinates of the center: ({cx}, {cy})")
                cv2.circle(image_copy, (int(x), int(y)), radius= 0, color = (255, 0, 0), thickness= -1)
    return image_copy, x, y
```

```python
#GET TWO POINT
import sys
sys.path.append("D:\\Quan\\roboDK\\Vision-Machine-collab-Nhan\\VIKO_UltraRobot")

from robodk.robolink import *  # API to communicate with RoboDK
from robodk.robomath import *  # basic matrix operations
from library import robot_lib as rl
from src import robotic_modify as RM

class Z_Axis:

    def __init__(self):
        self.rob_mod = RM.VisionRobot()
        self.rob_mod.pickRobot()
        self.rob_mod.fixedRef()
        self.rob_mod.attRobot()
        self.robot_module = rl.RobotModule()

    @staticmethod
    def sleep_seconds(seconds):
        print(f"Sleeping for {seconds} seconds...")
        time.sleep(seconds)
        print("Awake now!")
    

    def move(self):

        self.robot_module.cameraPosLeft(self.rob_mod.rf_laser2base)
        self.sleep_seconds(5)
        self.img1 = getImage('p1')
        self.robot_module.cameraPosRight(self.rob_mod.rf_laser2base)
        self.sleep_seconds(5)
        self.img2 = getImage('p2')
    

rot = Z_Axis()
rot.move()
```

```python
#CALCULATE THE DISTANCE
from config import config as CFG

def distanceCameraToObject(x1_dis, x2_dis, focal_length, baseLine)-> list:
    
    pixel_focalLength = focal_length * 1000 / CFG.PIXEL_SIZE 
    dis_cameraToObject = baseLine * pixel_focalLength / (abs(x1_dis - x2_dis))
    print(f'dis_cameraToObject:{dis_cameraToObject}')
    return dis_cameraToObject, pixel_focalLength

pixel_x1 = p1[1]
pixel_y1 = p1[2]
pixel_x2 = p2[1]
pixel_y2 = p2[2]

pixel_focalLength = 16 * 1000 / CFG.PIXEL_SIZE 
o1o2 = CFG.FORWARD_BASELINE
res_width = 2448
res_height = 2048

focal_length = 16
dis = distanceCameraToObject(pixel_y1, pixel_y2, focal_length, o1o2)
print(dis)
```
