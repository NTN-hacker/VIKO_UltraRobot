```python
import numpy as np
from collections import Counter

def getLabels(results):
    """
    Save data predict when infer
    """
    labels = []
    for result in results:
        boxes = result.boxes
        for box in boxes:
            label = int(box.cls.item())
            labels.append(label)
    return labels

def count_defects(defects):
    return dict(Counter(defects))

# Example usage
dict_re = model.predict(img_cvt)
out_labels = getLabels(dict_re)

# Convert labels to their respective defect names
ref_label = [DEFECT_LABEL[label] for label in out_labels]

# Count the defects
defects = np.array(ref_label)
result = count_defects(defects)

print(result)
```
